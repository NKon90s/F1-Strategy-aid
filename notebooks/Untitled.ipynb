```python
def add_missing_compounds(df):
    df = df.copy()

    compounds = {'SOFT', 'MEDIUM', 'HARD'}

    # Columns identifying a unique race and driver
    group_cols = ['meeting_name', 'driver_number', 'full_name']

    # Columns for which we want to calculate the mean
    mean_cols = ['driver_pace_by_comp', 'Avg_Deg_Rate']

    new_rows = []

    # Check each race-driver combination
    for _, group in df.groupby(group_cols, dropna=False):
        existing = set(group['compound'].dropna().unique())
        missing = compounds - existing

        for compound in missing:
            # Start from an existing row to preserve the schema
            row = group.iloc[0].copy()

            # Assign the missing compound
            row['compound'] = compound

            # Fill the two columns with their group means
            for col in mean_cols:
                row[col] = group[col].mean()

            new_rows.append(row)

    # Append all newly created rows
    if new_rows:
        df = pd.concat(
            [df, pd.DataFrame(new_rows)],
            ignore_index=True
        )

    return df
```
